```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime

# Set plotting style
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

print("Libraries imported successfully.")
```

```python
# Loading all datasets based on the schema provided
# Note: Assuming low_memory=False to prevent type mixing warnings on large CSVs

try:
    df_participants = pd.read_csv('participants.csv', low_memory=False)
    df_funnel = pd.read_csv('/content/engagement_events.csv', low_memory=False)
    df_studies = pd.read_csv('studies.csv', low_memory=False)
    df_submissions = pd.read_csv('submissions.csv', low_memory=False)
    df_payments = pd.read_csv('payments.csv', low_memory=False)
    df_events = pd.read_csv('/content/onboarding.csv', low_memory=False)

    print(" All datasets loaded successfully.")
    print(f"Participants: {df_participants.shape}")
    print(f"Funnel: {df_funnel.shape}")
    print(f"Studies: {df_studies.shape}")
    print(f"Submissions: {df_submissions.shape}")
    print(f"Payments: {df_payments.shape}")
    print(f"Events: {df_events.shape}")

except FileNotFoundError as e:
    print(f" Error loading files: {e}")
    print("Please ensure CSV files are in the current working directory.")
```

```python
datetime_cols = {
    'participants': ['signup_date', 'last_activity_date'],
    'funnel': ['signup_date', 'email_verified_date', 'profile_completed_date',
               'demographic_survey_completed_date', 'first_study_completed_date'],
    'studies': ['posted_date'],
    'submissions': ['submission_date'],
    'payments': ['payment_date'],
    'events': ['event_timestamp', 'signup_date'] # Added signup_date to events datetime conversion
}

# Helper function to convert dates
def convert_dates(df, cols):
    for col in cols:
        if col in df.columns:
            df[col] = pd.to_datetime(df[col], errors='coerce')
    return df

# Applying conversions
df_participants = convert_dates(df_participants, datetime_cols['participants'])
df_funnel = convert_dates(df_funnel, datetime_cols['funnel'])
df_studies = convert_dates(df_studies, datetime_cols['studies'])
df_submissions = convert_dates(df_submissions, datetime_cols['submissions'])
df_payments = convert_dates(df_payments, datetime_cols['payments'])
df_events = convert_dates(df_events, datetime_cols['events'])

print(" Date columns converted to datetime objects.")
print("\n--- Participants Info ---")
print(df_participants.info())
```

```python
# First, let's establish a "today" reference point.
# In production, this would be datetime.now(), but for this dataset,
# we'll use the max date in the data to keep it relative.
reference_date = df_participants['last_activity_date'].max()
print(f"Reference Date for Calculations: {reference_date}")

# 1. Aggregate Submissions per participant
sub_agg = df_submissions.groupby('participant_id').agg(
    total_submissions=('submission_id', 'count'),
    studies_completed=('status', lambda x: (x == 'completed').sum()),
    studies_rejected=('status', lambda x: (x == 'rejected').sum()),
    avg_completion_time=('completion_time_minutes', 'mean')
).reset_index()

# 2. Aggregate Payments per participant
pay_agg = df_payments.groupby('participant_id').agg(
    total_earnings=('amount', 'sum')
).reset_index()

# 3. Aggregate Events per participant (for engagement)
ev_agg = df_events.groupby('participant_id').size().reset_index(name='total_events')

# 4. Get first study completed date from funnel
# Correcting the DataFrame from df_funnel to df_events as these columns are in df_events
activation_data = df_events[['participant_id', 'signup_date', 'first_study_completed_date']].copy()
# Rename signup_date to avoid conflict during merge
activation_data.rename(columns={'signup_date': 'funnel_signup_date'}, inplace=True)

print(" Aggregations prepared.")
```

```python
# Start with the participants table as our base
df_features = df_participants.copy()

# Merge our aggregated data
df_features = df_features.merge(sub_agg, on='participant_id', how='left')
df_features = df_features.merge(pay_agg, on='participant_id', how='left')
df_features = df_features.merge(ev_agg, on='participant_id', how='left')
df_features = df_features.merge(activation_data, on='participant_id', how='left')

# Fill NaNs for users with no submissions, payments, or events
df_features[['total_submissions', 'studies_completed', 'studies_rejected']] = \
    df_features[['total_submissions', 'studies_completed', 'studies_rejected']].fillna(0)
df_features['total_earnings'] = df_features['total_earnings'].fillna(0)
df_features['total_events'] = df_features['total_events'].fillna(0)
df_features['avg_completion_time'] = df_features['avg_completion_time'].fillna(0)

print("Base feature matrix created. Shape:", df_features.shape)
df_features.head()
```

```python
# Ensure relevant columns are datetime objects, as they might have been merged as objects if not fully converted upstream.
# This explicitly handles potential type issues from prior merges.
df_features['funnel_signup_date'] = pd.to_datetime(df_features['funnel_signup_date'], errors='coerce')
df_features['first_study_completed_date'] = pd.to_datetime(df_features['first_study_completed_date'], errors='coerce')

# 1. Account Age (Days since signup)
df_features['account_age_days'] = (reference_date - df_features['signup_date']).dt.days

# 2. Days Since Signup (Identical to Account Age, but explicitly requested.
# Sometimes used as a windowing function, but here it's absolute days)
df_features['days_since_signup'] = (reference_date - df_features['signup_date']).dt.days

# 3. Days Since Last Activity
df_features['days_since_last_activity'] = (reference_date - df_features['last_activity_date']).dt.days

# 4. Time to Activation (Days from signup to first completed study)
# If not activated, leave as NaT, then fill with a large number or -1 for modeling later
df_features['time_to_activation_days'] = (
    df_features['first_study_completed_date'] - df_features['funnel_signup_date']
).dt.days

# 5. Acquisition Cohort (Month-Year of Signup)
df_features['acquisition_cohort'] = df_features['signup_date'].dt.to_period('M').astype(str)

print("Time-based features engineered.")
df_features[['participant_id', 'account_age_days', 'days_since_last_activity',
             'time_to_activation_days', 'acquisition_cohort']].head()
```

```python
# 6. Studies Completed (Already merged, just ensuring int type)
df_features['studies_completed'] = df_features['studies_completed'].astype(int)

# 7. Total Earnings (Already merged)
# 8. Average Reward per Study
df_features['avg_reward_per_study'] = np.where(
    df_features['studies_completed'] > 0,
    df_features['total_earnings'] / df_features['studies_completed'],
    0
)

# 9. Average Completion Time (Already merged, but let's filter out 0s for better avg if needed.
# Current logic includes 0s for those who never did a study. Let's keep it as is for now.)

# 10. Activity Frequency (Submissions per week since signup)
# Add 1 to account_age to avoid division by zero for users who signed up today
df_features['activity_frequency'] = df_features['total_submissions'] / ((df_features['account_age_days'] + 1) / 7)

print(" Behavioral & Financial features engineered.")
df_features[['participant_id', 'studies_completed', 'total_earnings',
             'avg_reward_per_study', 'activity_frequency']].head()
```

```python
# 11. Participant Quality Score
# Logic: Completion Rate (Completed / Total Submissions) * 100.
# Penalize slightly for rejections.
df_features['completion_rate'] = np.where(
    df_features['total_submissions'] > 0,
    df_features['studies_completed'] / df_features['total_submissions'],
    0
)
df_features['rejection_rate'] = np.where(
    df_features['total_submissions'] > 0,
    df_features['studies_rejected'] / df_features['total_submissions'],
    0
)
# Quality Score = (Completion Rate * 100) - (Rejection Rate * 20)
df_features['participant_quality_score'] = (df_features['completion_rate'] * 100) - (df_features['rejection_rate'] * 20)
df_features['participant_quality_score'] = df_features['participant_quality_score'].clip(lower=0, upper=100)

# 12. Churn Flag
# Based on schema: churned (>90d). We also check 'at_risk' (31-90d) but flag specifically for churned.
df_features['churn_flag'] = np.where(df_features['days_since_last_activity'] > 90, 1, 0)

# 13. Reactivation Flag
# Logic: User was inactive for >90 days (churned) but has recent activity (<=30 days)
# We check if account_age is > 90, days_since_last_activity is low, but they had a gap.
# A simpler proxy based on available data: account_status is 'active' but account_age > 90 and they have > 0 studies completed.
# Let's use the event data to see if they had a gap.
# Since we only have aggregated events, let's define it as:
# Active now (<=30d since activity) BUT account age is >90d AND total submissions are low relative to account age.
# Better approach based on provided schema:
df_features['reactivation_flag'] = np.where(
    (df_features['days_since_last_activity'] <= 30) &
    (df_features['account_age_days'] > 90) &
    (df_features['activity_frequency'] < 0.5), # Sporadic historical activity
    1,
    0
)

print(" Scoring & Flag features engineered.")
df_features[['participant_id', 'participant_quality_score', 'churn_flag', 'reactivation_flag']].head()
```

```python
# 14. Engagement Score (0-100 scale)
# Components: Recent Activity (40%), Study Volume (30%), Event Interaction (30%)
# We will use Min-Max scaling concept manually.

# Recency score: 100 if active today, 0 if >90 days. Linear decay.
df_features['recency_score'] = 100 - (df_features['days_since_last_activity'].clip(upper=90) * (100/90))

# Volume score: Min-Max scale studies completed (capped at 20 studies for 100 pts)
df_features['volume_score'] = (df_features['studies_completed'].clip(upper=20) / 20) * 100

# Interaction score: Min-Max scale events (capped at 100 events for 100 pts)
df_features['interaction_score'] = (df_features['total_events'].clip(upper=100) / 100) * 100

# Final Weighted Engagement Score
df_features['engagement_score'] = (
    (df_features['recency_score'] * 0.4) +
    (df_features['volume_score'] * 0.3) +
    (df_features['interaction_score'] * 0.3)
).round(2)

# 15. Lifetime Value (LTV)
# For a research participant, LTV is essentially their total earnings to date
# PLUS a predicted future value based on engagement.
# Predicted future value = (Engagement Score / 100) * Average Monthly Earning
# To keep it grounded in current data: LTV = Total Earnings * (1 + (Engagement Score/100))
# This assumes highly engaged users will bring in up to 100% more of their current total.
df_features['ltv'] = (df_features['total_earnings'] * (1 + (df_features['engagement_score'] / 100))).round(2)

# Clean up intermediate columns used for scoring
df_features.drop(['recency_score', 'volume_score', 'interaction_score',
                  'completion_rate', 'rejection_rate'], axis=1, inplace=True)

print(" Engagement Score & LTV engineered.")
```

```python
# Display the final feature set
final_cols = [
    'participant_id', 'acquisition_cohort', 'account_age_days', 'days_since_signup',
    'days_since_last_activity', 'time_to_activation_days', 'studies_completed',
    'total_earnings', 'avg_reward_per_study', 'avg_completion_time', 'activity_frequency',
    'engagement_score', 'participant_quality_score', 'churn_flag', 'reactivation_flag', 'ltv',
    'account_status' # Added account_status to final features
]

df_final_features = df_features[final_cols]

print(f"Final Feature Matrix Shape: {df_final_features.shape}")
print("\n--- Sample of Final Features ---")
df_final_features.head(10).style.background_gradient(cmap='viridis', subset=['engagement_score', 'participant_quality_score', 'ltv'])
```

```python
# Merge submissions with participant signup dates
df_retention = df_submissions.merge(df_participants[['participant_id', 'signup_date']], on='participant_id')

# Extract Year-Month for signup and submission
df_retention['signup_month'] = df_retention['signup_date'].dt.to_period('M')
df_retention['active_month'] = df_retention['submission_date'].dt.to_period('M')

# Calculate month difference (Cohort Index)
df_retention['cohort_index'] = (df_retention['active_month'] - df_retention['signup_month']).apply(lambda x: x.n)

# Get unique active users per cohort per month index
cohort_data = df_retention.groupby(['signup_month', 'cohort_index'])['participant_id'].nunique().reset_index()

# Pivot to create the retention matrix
cohort_counts = cohort_data.pivot(index='signup_month', columns='cohort_index', values='participant_id')

# Get cohort sizes (total signups per month)
cohort_sizes = df_participants.groupby(df_participants['signup_date'].dt.to_period('M'))['participant_id'].nunique()

# Calculate retention percentages
retention_matrix = cohort_counts.divide(cohort_sizes, axis=0) * 100

# Plot the retention matrix as a heatmap
plt.figure(figsize=(14, 8))
sns.heatmap(retention_matrix, annot=True, fmt='.1f', cmap='Blues', vmin=0, vmax=50)
plt.title('Monthly Retention Cohorts (% of Cohort Active via Submissions)')
plt.xlabel('Months Since Signup')
plt.ylabel('Acquisition Cohort Month')
plt.show()
```

```python
# We will use the df_final_features matrix from Phase 2,
# plus pull in account_status from df_participants for precise schema definitions.

df_segments = df_final_features.copy()
df_segments = df_segments.merge(df_participants[['participant_id', 'account_status']], on='participant_id', how='left')

# Calculate dynamic thresholds for "Power" and "High Earner" segments
high_earner_threshold = df_segments['total_earnings'].quantile(0.90)
power_user_studies_threshold = df_segments['studies_completed'].quantile(0.75)

print(f"High Earner Threshold (Top 10%): > ${high_earner_threshold:.2f}")
print(f"Power User Threshold (Top 25%): > {power_user_studies_threshold:.0f} studies completed")

def assign_segment(row):
    # 1. Lifecycle states take precedence
    if row['churn_flag'] == 1:
        return 'Churned Users'
    if row['reactivation_flag'] == 1:
        return 'Reactivated Users'

    # 2. New vs Dormant (Based on Account Age <= 14d onboarding window)
    if row['account_age_days'] <= 14 and row['studies_completed'] == 0:
        return 'New Users'
    if row['studies_completed'] == 0 and row['account_age_days'] > 14:
        return 'Dormant Users' # Signed up but never activated and past onboarding

    # 3. At-Risk (Active in 31-90d window)
    if 31 <= row['days_since_last_activity'] <= 90:
        return 'At-Risk Users'

    # 4. Valued Segments (Currently Active <= 30d)
    if row['days_since_last_activity'] <= 30:
        if row['total_earnings'] >= high_earner_threshold and row['total_earnings'] > 0:
            return 'High Earners'
        if row['studies_completed'] >= power_user_studies_threshold and row['studies_completed'] > 0:
            return 'Power Users'
        return 'Active Users'

    # Fallback (e.g., edge cases in timing)
    return 'Active Users'

# Apply segmentation
df_segments['segment'] = df_segments.apply(assign_segment, axis=1)

print(" Segmentation logic applied.")
```

```python
from datetime import timedelta

def calculate_kpis(participants, funnel, submissions, events, payments, as_of_date):
    """
    Calculates platform KPIs as of a specific date.
    Assumes dataframes are pre-filtered or contains all historical data up to as_of_date.
    """
    kpis = {}

    # Time windows
    date_d = as_of_date
    date_w = as_of_date - timedelta(days=7)
    date_m = as_of_date - timedelta(days=30)
    date_90d = as_of_date - timedelta(days=90)

    # 1. Acquisition (Signups in last 30 days)
    kpis['acquisition_30d'] = participants[participants['signup_date'] >= date_m]['participant_id'].nunique()

    # 2. Activation Rate (Overall: Activated Users / Total Signups)
    total_signups = len(funnel)
    activated_users = funnel['is_activated'].sum()
    kpis['activation_rate'] = (activated_users / total_signups * 100) if total_signups > 0 else 0

    # Active Users base (using events as the pulse of activity)
    active_d = events[events['event_timestamp'].dt.date == date_d.date()]['participant_id'].nunique()
    active_w = events[(events['event_timestamp'] >= date_w) & (events['event_timestamp'] <= date_d)]['participant_id'].nunique()
    active_m = events[(events['event_timestamp'] >= date_m) & (events['event_timestamp'] <= date_d)]['participant_id'].nunique()

    # 3, 4, 5. DAU, WAU, MAU
    kpis['dau'] = active_d
    kpis['wau'] = active_w
    kpis['mau'] = active_m

    # 6. Stickiness Ratio (DAU/MAU)
    kpis['stickiness_ratio'] = (active_d / active_m * 100) if active_m > 0 else 0

    # 7. Retention Rate (W4 Retention: Users active in month 1 who are also active in month 0-1 window)
    # Simplified for snapshot: % of MAU who were also active in the preceding 30 days
    prev_m_active = events[(events['event_timestamp'] >= date_m - timedelta(days=30)) &
                           (events['event_timestamp'] < date_m)]['participant_id'].unique()
    curr_m_active = events[(events['event_timestamp'] >= date_m) &
                           (events['event_timestamp'] <= date_d)]['participant_id'].unique()
    if len(prev_m_active) > 0:
        retained = len(set(prev_m_active).intersection(set(curr_m_active)))
        kpis['retention_rate'] = (retained / len(prev_m_active) * 100)
    else:
        kpis['retention_rate'] = 0

    # 8. Churn Rate (% of prev month active who did NOT return this month)
    if len(prev_m_active) > 0:
        churned = len(set(prev_m_active) - set(curr_m_active))
        kpis['churn_rate'] = (churned / len(prev_m_active) * 100)
    else:
        kpis['churn_rate'] = 0

    # 9. Reactivation Rate (Users active this month who were NOT active last month, but were active before)
    all_time_before_m = events[events['event_timestamp'] < date_m]['participant_id'].unique()
    reactivated = len(set(curr_m_active) - set(prev_m_active).union(set(all_time_before_m)))
    # Actually, reactivated = Active this month, NOT active last month, BUT active sometime before last month
    true_reactivated = set(curr_m_active) - set(prev_m_active)
    true_reactivated = true_reactivated.intersection(set(all_time_before_m))
    kpis['reactivation_rate'] = (len(true_reactivated) / len(curr_m_active) * 100) if len(curr_m_active) > 0 else 0

    # Financial Metrics (30-day window)
    subs_30d = submissions[(submissions['submission_date'] >= date_m) & (submissions['submission_date'] <= date_d)]

    # 10. Average Reward (Completed studies in last 30d)
    completed_30d = subs_30d[subs_30d['status'] == 'completed']
    kpis['avg_reward'] = completed_30d['reward_earned'].mean() if len(completed_30d) > 0 else 0

    # 11. Total Revenue (Total payouts in last 30d)
    pay_30d = payments[(payments['payment_date'] >= date_m) & (payments['payment_date'] <= date_d)]
    kpis['total_revenue_30d'] = pay_30d['amount'].sum()

    # 12. LTV (Average lifetime earnings per activated user)
    # Using total dataset for lifetime averages
    kpis['ltv'] = (participants['total_earnings'].sum() / activated_users) if activated_users > 0 else 0

    # 13. Participant Quality (Avg quality score of active users this month)
    active_quality = participants[participants['participant_id'].isin(curr_m_active)]['participant_quality_score'].mean()
    kpis['participant_quality'] = active_quality if not pd.isna(active_quality) else 0

    # 14. Marketplace Health Score (0-100 Composite)
    # Formula: 30% Activation + 30% Stickiness + 20% Quality + 20% Retention
    # All components naturally scale 0-100
    health_score = (
        (kpis['activation_rate'] * 0.3) +
        (kpis['stickiness_ratio'] * 0.3) +
        (kpis['participant_quality'] * 0.2) +
        (kpis['retention_rate'] * 0.2)
    )
    kpis['marketplace_health_score'] = round(health_score, 1)

    return kpis

print(" KPI Function defined successfully.")
```

```python
# Question: Do Academic Research studies take significantly longer to complete than Commercial studies?
# H0: There is no difference in completion time between Academic and Commercial studies.
# H1: Academic studies take significantly longer.

academic_subs = df_submissions.merge(df_studies[['study_id', 'researcher_type']], on='study_id')
academic_times = academic_subs[academic_subs['researcher_type'] == 'Academic']['completion_time_minutes'].dropna()
commercial_times = academic_subs[academic_subs['researcher_type'] == 'Commercial']['completion_time_minutes'].dropna()

# Filter out extreme outliers (>180 mins) for a fair test
academic_times = academic_times[academic_times <= 180]
commercial_times = commercial_times[commercial_times <= 180]

# Perform Independent T-Test (Welch's t-test since variances might be unequal)
t_stat, p_value = stats.ttest_ind(academic_times, commercial_times, equal_var=False)

print("--- Hypothesis Test: Completion Time (Academic vs Commercial) ---")
print(f"Mean Academic Time: {academic_times.mean():.2f} mins")
print(f"Mean Commercial Time: {commercial_times.mean():.2f} mins")
print(f"T-Statistic: {t_stat:.2f}")
print(f"P-Value: {p_value:.4e}")

alpha = 0.05
if p_value < alpha:
    print("➡️ Result: Reject the null hypothesis. There is a statistically significant difference in completion times.")
else:
    print("➡️ Result: Fail to reject the null hypothesis. No significant difference found.")
```

```python
# Let's calculate the 95% Confidence Interval for the Average Hourly Rate earned by participants.
# This helps us understand the range in which the true population mean lies.

completed = df_submissions[df_submissions['status'] == 'completed'].copy()
completed['implied_hourly_rate'] = (completed['reward_earned'] / completed['completion_time_minutes']) * 60

# Filter out nonsense rates (e.g., 0 min completion times causing inf rates, or > $100/hr)
valid_rates = completed[(completed['implied_hourly_rate'] > 0) & (completed['implied_hourly_rate'] < 100)]['implied_hourly_rate']

sample_mean = valid_rates.mean()
sample_std = valid_rates.std()
n = len(valid_rates)
standard_error = sample_std / np.sqrt(n)

# 95% CI Z-score is ~1.96
z_score = 1.96
margin_of_error = z_score * standard_error

lower_bound = sample_mean - margin_of_error
upper_bound = sample_mean + margin_of_error

print("--- 95% Confidence Interval: Implied Hourly Rate ---")
print(f"Sample Size: {n}")
print(f"Mean Hourly Rate: ${sample_mean:.2f}")
print(f"95% CI: [${lower_bound:.2f}, ${upper_bound:.2f}]")
print("\nInterpretation: We are 95% confident that the true average hourly rate for all participants falls within this range.")
print("Platform Target is ~$10.50/hr. The target is within our confidence interval, but skewed toward the lower end.")
```

```python
# Scenario: We A/B tested a new "Quick Match" study recommendation feature.
# Group A (Control): Standard study feed.
# Group B (Treatment): Quick Match feed.
# Metric: 7-day Activation Rate (from signup to first completed study).

np.random.seed(42) # For reproducibility

# Simulating A/B groups since we don't have an explicit experiment flag
df_ab = df_funnel.sample(n=10000, random_state=42).copy()
df_ab['group'] = np.random.choice(['A (Control)', 'B (Treatment)'], size=len(df_ab), p=[0.5, 0.5])

# Simulate a treatment effect: Group B gets a 4% absolute lift in activation
def simulate_activation(row):
    if row['group'] == 'A (Control)':
        return 1 if np.random.rand() < 0.20 else 0  # 20% baseline activation
    else:
        return 1 if np.random.rand() < 0.24 else 0  # 24% treatment activation

df_ab['sim_activated'] = df_ab.apply(simulate_activation, axis=1)

# Calculate Results
results = df_ab.groupby('group')['sim_activated'].agg(['count', 'sum'])
results['rate'] = results['sum'] / results['count']

print("--- A/B Test Results: 7-Day Activation ---")
print(results)

# Perform Two-Proportion Z-Test
from statsmodels.stats.proportion import proportions_ztest, proportion_confint

count = results['sum'].values
nobs = results['count'].values

z_stat, p_val = proportions_ztest(count, nobs)
(lower_ci, upper_ci) = proportion_confint(count, nobs, alpha=0.05)

print(f"\nZ-Statistic: {z_stat:.2f}")
print(f"P-Value: {p_val:.4f}")

if p_val < 0.05:
    print("Result: Statistically significant. Roll out Quick Match feature to 100%.")
else:
    print(" Result: Not significant. Do not roll out.")

print(f"\n95% CI for Activation Rate:")
print(f"Group A: [{lower_ci[0]*100:.2f}%, {upper_ci[0]*100:.2f}%]")
print(f"Group B: [{lower_ci[1]*100:.2f}%, {upper_ci[1]*100:.2f}%]")
```
